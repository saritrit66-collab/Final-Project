```python
import pandas as pd
import time
import re
from urllib.parse import urlparse
from tqdm import tqdm
import google_colab_selenium as gs

# פונקציית עזר לתרגום טקסט למספרים (למשל 1.5K -> 1500)
def parse_number(num_str):
    if not num_str: return 0
    num_str = str(num_str).lower().replace(',', '').strip()
    try:
        if 'k' in num_str: return int(float(num_str.replace('k', '')) * 1000)
        if 'm' in num_str: return int(float(num_str.replace('m', '')) * 1000000)
        return int(float(num_str))
    except:
        return 0

# הגדרת הדפדפן המיוחד של קולאב
print("פותח דפדפן סמוי (זה ייקח כמה שניות)...")
driver = gs.Chrome()
driver.set_page_load_timeout(15)

# פונקציית הנתב החכם (Smart Router)
def smart_scrape(url):
    if pd.isna(url) or not isinstance(url, str) or not url.startswith('http'):
        return 0, 0, 0

    likes, shares, comments = 0, 0, 0
    domain = urlparse(url).netloc.lower()

    try:
        driver.get(url)
        time.sleep(4) # המתנה קריטית לטעינת המספרים
        page_text = driver.page_source.lower()

        # --- לוגיקה 1: יוטיוב ---
        if 'youtube.com' in domain or 'youtu.be' in domain:
            # ביוטיוב הלייקים לרוב מופיעים ליד המילה like או באזור כפתור הלייק
            likes_match = re.search(r'([\d\.,km]+)\s*likes?', page_text)
            comments_match = re.search(r'([\d\.,km]+)\s*comments?', page_text)
            if likes_match: likes = parse_number(likes_match.group(1))
            if comments_match: comments = parse_number(comments_match.group(1))

        # --- לוגיקה 2: טוויטר / X (ניסיון עקיפה) ---
        elif 'twitter.com' in domain or 'x.com' in domain:
            # בטוויטר המילים הן Reposts, Quotes, Likes
            likes_match = re.search(r'([\d\.,km]+)\s*(?:likes?|אהבתי)', page_text)
            shares_match = re.search(r'([\d\.,km]+)\s*(?:reposts?|retweets?|quotes?)', page_text)
            comments_match = re.search(r'([\d\.,km]+)\s*(?:replies?|תגובות)', page_text)

            if likes_match: likes = parse_number(likes_match.group(1))
            if shares_match: shares = parse_number(shares_match.group(1))
            if comments_match: comments = parse_number(comments_match.group(1))

        # --- לוגיקה 3: כל אתר אחר (חדשות, בלוגים) ---
        else:
            likes_match = re.search(r'([\d\.,km]+)\s*(?:likes?|upvotes?|אהבו)', page_text)
            shares_match = re.search(r'([\d\.,km]+)\s*(?:shares?|שיתופים)', page_text)
            comments_match = re.search(r'([\d\.,km]+)\s*(?:comments?|replies?|תגובות)', page_text)

            if likes_match: likes = parse_number(likes_match.group(1))
            if shares_match: shares = parse_number(shares_match.group(1))
            if comments_match: comments = parse_number(comments_match.group(1))

    except Exception as e:
        pass # דילוג שקט במקרה של חסימה כדי שהלולאה לא תקרוס

    return likes, shares, comments

# --- הפעלת הסריקה ---
# ! חשוב: החליפי את שם הקובץ לשם האמיתי שלך !
# df = pd.read_csv('your_file_name.csv', encoding='utf-8', encoding_errors='ignore')

# ניקח דגימה של 20 שורות לבדיקה
df_sample = df.head(20).copy()

df_sample['Scraped_Likes'] = 0
df_sample['Scraped_Shares'] = 0
df_sample['Scraped_Comments'] = 0

print(f"מתחיל סריקה של {len(df_sample)} לינקים...")
for index, row in tqdm(df_sample.iterrows(), total=df_sample.shape[0]):
    l, s, c = smart_scrape(row['URL'])
    df_sample.at[index, 'Scraped_Likes'] = l
    df_sample.at[index, 'Scraped_Shares'] = s
    df_sample.at[index, 'Scraped_Comments'] = c

driver.quit()

print("\nהסריקה הסתיימה. תוצאות:")
print(df_sample[['URL', 'Scraped_Likes', 'Scraped_Shares', 'Scraped_Comments']])
```
