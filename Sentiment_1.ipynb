```python
import pandas as pd
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
from tqdm.notebook import tqdm

# אתחול
analyzer = SentimentIntensityAnalyzer()

def get_vader_sentiment(text):
    if pd.isna(text) or text == "":
        return 0.0
    return analyzer.polarity_scores(str(text))['compound']

# שם הקובץ שהעלית (וודא שהשם מדויק כולל הסיומת)
file_name = 'your_data.csv'

print("✅ המערכת מוכנה לניתוח")
```

```python
try:
    # טעינת הקובץ מהתיקייה הנוכחית
    df = pd.read_csv(file_name, encoding='utf-8-sig')
    print(f"הקובץ נטען! נמצאו {len(df)} שורות.")

    # וודא ששם העמודה כאן תואם לשם העמודה בקובץ שלך
    column_to_analyze = 'text'

    tqdm.pandas()
    print("מנתח סנטימנט...")
    df['sentiment_score'] = df[column_to_analyze].progress_apply(get_vader_sentiment)

    # הוספת תווית (חיובי/שלילי/נייטרלי)
    df['sentiment_label'] = df['sentiment_score'].apply(
        lambda c: 'Positive' if c >= 0.05 else ('Negative' if c <= -0.05 else 'Neutral')
    )

    # הצגת תוצאות ראשוניות
    display(df.head())

except FileNotFoundError:
    print(f"❌ שגיאה: הקובץ '{file_name}' לא נמצא. וודא שהעלית אותו למחברת ושמו תואם.")
except Exception as e:
    print(f"❌ קרתה שגיאה: {e}")
```

```python
output_name = 'sentiment_results.csv'
df.to_csv(output_name, index=False, encoding='utf-8-sig')

print(f"✨ הסתיים! הקובץ '{output_name}' נוצר בתיקיית המחברת. ניתן להוריד אותו כעת.")
```
